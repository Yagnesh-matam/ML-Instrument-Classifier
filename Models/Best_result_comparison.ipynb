```python
import matplotlib.pyplot as plt
import numpy as np

# Data from model evaluations
models_data = [
    {'model': 'Adaboost', 'accuracy': 0.4183, 'precision': 0.4293, 'f1_score': 0.4162, 'recall': 0.4183},
    {'model': 'ANN', 'accuracy': 0.50, 'precision': 0.50, 'f1_score': 0.49, 'recall': 0.50},
    {'model': 'Decision_Tree', 'accuracy': 0.4131, 'precision': 0.4201184502095047, 'f1_score': 0.4150558694537467, 'recall': 0.41312453392990306},
    {'model': 'Gradient_Boost', 'accuracy': 0.5183, 'precision': 0.5255, 'f1_score': 0.5177, 'recall': 0.5183},
    {'model': 'KNN', 'accuracy': 0.6413, 'precision': 0.66406, 'f1_score': 0.64036, 'recall': 0.64131},
    {'model': 'Logistic_Regression', 'accuracy': 0.5802, 'precision': 0.58163764, 'f1_score': 0.57879, 'recall': 0.58016},
    {'model': 'Naive_Bayes', 'accuracy': 0.3177, 'precision': 0.3351, 'f1_score': 0.2856, 'recall': 0.3177},
    {'model': 'Random_Forest', 'accuracy': 0.6339, 'precision': 0.6448, 'f1_score': 0.6317, 'recall': 0.6338},
    {'model': 'SVM', 'accuracy': 0.6443, 'precision': 0.6463, 'f1_score': 0.6419, 'recall': 0.6443},
]

# Extract model names and metrics
models = [d['model'] for d in models_data]
accuracies = [d['accuracy'] for d in models_data]
precisions = [d['precision'] for d in models_data]
f1_scores = [d['f1_score'] for d in models_data]
recalls = [d['recall'] for d in models_data]
```

```python
# Plot Accuracy
plt.figure(figsize=(12, 6))
bars = plt.bar(models, accuracies, color='skyblue')
plt.title('Model Accuracy Comparison')
plt.xlabel('Models')
plt.ylabel('Accuracy')
plt.xticks(rotation=45, ha='right')
plt.ylim(0, 1)
for bar, acc in zip(bars, accuracies):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01, f'{acc:.4f}', ha='center', va='bottom')
plt.tight_layout()
plt.show()

# Best model for Accuracy
best_acc_idx = np.argmax(accuracies)
best_acc_model = models[best_acc_idx]
best_acc_value = accuracies[best_acc_idx]
print(f"The best model according to Accuracy is {best_acc_model} with a score of {best_acc_value:.4f}")
```

```python
# Plot Precision
plt.figure(figsize=(12, 6))
bars = plt.bar(models, precisions, color='lightgreen')
plt.title('Model Precision Comparison')
plt.xlabel('Models')
plt.ylabel('Precision')
plt.xticks(rotation=45, ha='right')
plt.ylim(0, 1)
for bar, prec in zip(bars, precisions):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01, f'{prec:.4f}', ha='center', va='bottom')
plt.tight_layout()
plt.show()

# Best model for Precision
best_prec_idx = np.argmax(precisions)
best_prec_model = models[best_prec_idx]
best_prec_value = precisions[best_prec_idx]
print(f"The best model according to Precision is {best_prec_model} with a score of {best_prec_value:.4f}")
```

```python
# Plot F1 Score
plt.figure(figsize=(12, 6))
bars = plt.bar(models, f1_scores, color='lightcoral')
plt.title('Model F1 Score Comparison')
plt.xlabel('Models')
plt.ylabel('F1 Score')
plt.xticks(rotation=45, ha='right')
plt.ylim(0, 1)
for bar, f1 in zip(bars, f1_scores):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01, f'{f1:.4f}', ha='center', va='bottom')
plt.tight_layout()
plt.show()

# Best model for F1 Score
best_f1_idx = np.argmax(f1_scores)
best_f1_model = models[best_f1_idx]
best_f1_value = f1_scores[best_f1_idx]
print(f"The best model according to F1 Score is {best_f1_model} with a score of {best_f1_value:.4f}")
```

```python
# Plot Recall
plt.figure(figsize=(12, 6))
bars = plt.bar(models, recalls, color='gold')
plt.title('Model Recall Comparison')
plt.xlabel('Models')
plt.ylabel('Recall')
plt.xticks(rotation=45, ha='right')
plt.ylim(0, 1)
for bar, rec in zip(bars, recalls):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01, f'{rec:.4f}', ha='center', va='bottom')
plt.tight_layout()
plt.show()

# Best model for Recall
best_rec_idx = np.argmax(recalls)
best_rec_model = models[best_rec_idx]
best_rec_value = recalls[best_rec_idx]
print(f"The best model according to Recall is {best_rec_model} with a score of {best_rec_value:.4f}")
```
